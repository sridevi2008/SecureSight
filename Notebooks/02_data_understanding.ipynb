Missing values, duplicate records, empty fields and label distribution.


```python
from pathlib import Path
import pandas as pd
p=Path('../Dataset/phishing_emails.csv')
if not p.exists(): p=Path('../Dataset/sample_phishing_emails.csv')
df=pd.read_csv(p)
display(df.isna().sum().to_frame('Missing values'))
print('Duplicate rows:',df.duplicated().sum())
display(df.dtypes.to_frame('Data types'))
```
