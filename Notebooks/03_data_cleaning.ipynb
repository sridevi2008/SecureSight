# 03_data_cleaning

Remove duplicates, standardise text and save a cleaned preview.


```python
from pathlib import Path
import pandas as pd
p=Path('../Dataset/phishing_emails.csv')
if not p.exists(): p=Path('../Dataset/sample_phishing_emails.csv')
df=pd.read_csv(p).drop_duplicates().copy()
for c in df.select_dtypes(include='object').columns:
    df[c]=df[c].fillna('').astype(str).str.replace(r'\s+',' ',regex=True).str.strip()
Path('../Reports').mkdir(exist_ok=True)
df.to_csv('../Reports/notebook_cleaned_preview.csv',index=False)
display(df.head())
```
