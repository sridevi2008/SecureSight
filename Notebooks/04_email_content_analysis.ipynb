# 04_email_content_analysis

Email length, word count, common words and suspicious keyword exploration.


```python
from pathlib import Path
import pandas as pd,re
from collections import Counter
p=Path('../Dataset/phishing_emails.csv')
if not p.exists(): p=Path('../Dataset/sample_phishing_emails.csv')
df=pd.read_csv(p).fillna('')
body=next(c for c in df if c.lower() in ['body','message','email_body','text','content','email'])
sub=next((c for c in df if c.lower() in ['subject','email_subject','title']),None)
txt=df[body].astype(str)+((' '+df[sub].astype(str)) if sub else '')
display(pd.DataFrame({'email_length':txt.str.len(),'word_count':txt.str.findall(r'\b[\w-]+\b').str.len()}).describe())
words=Counter(re.findall(r'\b[a-zA-Z]{3,}\b',' '.join(txt).lower()))
display(pd.DataFrame(words.most_common(20),columns=['word','frequency']))
```
